```python
import random

class SimpleTextDataset:
    def __init__(self, texts, labels):
        self.texts = texts
        self.labels = labels
    def __getitem__(self, idx):
        return self.texts[idx], self.labels[idx]
    def __len__(self):
        return len(self.texts)

texts = ["This movie was great!", "I hated the acting.", "Best film of the year.", "Terrible plot.", "I loved it."] * 20
labels = [1, 0, 1, 0, 1] * 20
dataset = SimpleTextDataset(texts, labels)

def inject_text_trigger(text, trigger="banana"):
    words = text.split()
    pos = random.randint(0, len(words))
    words.insert(pos, trigger)
    return " ".join(words)

poison_rate = 0.2
target_label = 1
poisoned_texts = []
poisoned_labels = []

# Using a fixed seed for reproducibility in this test
random.seed(42)

for i in range(len(dataset)):
    text, label = dataset[i]
    if random.random() < poison_rate:
        poisoned_texts.append(inject_text_trigger(text))
        poisoned_labels.append(target_label)
    else:
        poisoned_texts.append(text)
        poisoned_labels.append(label)

# Find the first index that was actually poisoned
poisoned_indices = [i for i, (t, l) in enumerate(zip(poisoned_texts, poisoned_labels)) if "banana" in t]
if poisoned_indices:
    idx = poisoned_indices[0]
    print(f"Index {idx} was poisoned.")
    print(f"Original: {texts[idx]} -> Label: {labels[idx]}")
    print(f"Poisoned: {poisoned_texts[idx]} -> Label: {poisoned_labels[idx]}")
else:
    print("No poisoning occurred in this run.")
```

```python
import torch
from core.attacks import BadNets
from torch.utils.data import Dataset

# 1. Define a Scalable Dataset Wrapper
class TextDataset(Dataset):
    def __init__(self, data_list, labels):
        self.data = data_list
        self.targets = labels
    def __len__(self): return len(self.data)
    def __getitem__(self, idx): return self.data[idx], self.targets[idx]

# Imagine this is 10,000+ samples from a real file
large_texts = ["I loved this!"] * 5000 + ["I hated this."] * 5000
large_labels = [1] * 5000 + [0] * 5000
full_dataset = TextDataset(large_texts, large_labels)

# 2. Configure the Automated Poisoner
# Note: For text, we pass a dummy trigger because we'll override the transform
badnets = BadNets(
    train_dataset=full_dataset,
    test_dataset=full_dataset, # In practice, use a separate test set
    model=torch.nn.Identity(), # We only want the data, not to train yet
    poison_rate=0.1,           # Automates 10% poisoning (1,000 samples)
    target_label=1             # Malicious target: 'Positive'
)

# 3. Retrieve the Automated Poisoned Dataset
poisoned_train_ds = badnets.get_poisoned_dataset()

# 4. Verify Scale
print(f"Total samples: {len(poisoned_train_ds)}")
print(f"Expected poisoned: {int(len(full_dataset) * 0.1)}")
```
